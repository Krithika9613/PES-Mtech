## Binomial Distribution

### A bank issues credit cards to customers under the scheme of Master Card.Based on the past Data, the bank found out that 60% of all accounts pay on time following the bill.If sample of 7 accounts selected at random from the current database,construct the Binomial probability Distribution of accounts paying on time.

```python
n=7
p=0.6  #for each account probability of success is 60% that is 0.6
```

```python
 k=np.arange(0,8)   # Since 7 account pay on time here we are creating array of 0  to 8. so either 0 pays on time  or all 7 pays on time or any o 
k
```

```python
binomial =stats.binom.pmf(k,n,p)   #binom.pmf stands for " Binomial Probability Mass Function"
binomial 
```

```python
# To see the distribution we can plot it using plot function
plt.plot(k,binomial,'o-')
plt.title("Binomial: n=%i,p=%.2f" % (n,p),fontsize=15)
plt.xlabel('Number of Successes')
plt.ylabel('probability of Successes')
plt.show()
```

```python
# how many Probability are calculated
n=len(binomial)
n
```

```python
# sum of the probability is 1 
sum(binomial)
```

## Poisson Distribution

### If on an average, 6 customers arrive every two minutes at a bank during the busy hours of working,
### a) What is the probability that exactly four customers arrive in a given minute? 
### b)What is the prrobability that more than three customers will arrive in a given minute?

```python
rate =3 # since 6 customer arrives at 2 minute .so 6/2=3 arrives at every 1 minute
n=np.arange(0,20) # Probability could be 0,1,2,3 .... so on.
                  # Given that average of 3 arrives it is extremely unlikely that is more than 20 customers arrives with in a minute.
                  # just to keep our array bounded we specify the probability from 0 to 20
```

```python
n
```

```python
poisson = stats.poisson.pmf(n,rate)
poisson
```

```python
# Probability of exactly 4 customer arrive in a given minute
poisson[4]
```

```python
#probability of more than 3 customers arrive in a minute 
1-(poisson[0]+poisson[1]+poisson[2]+poisson[3])
```

```python
plt.plot(n,poisson,'o-')
plt.title('Poisson: $\lambda$ = %i' %rate)
plt.xlabel("number of customer arriving in a minute")
plt.ylabel("Probability of number of customer arriving in a minute")
plt.show()
```

## Normal Distribution

### The mean weight of a morning breakfast cereal pack is 0.295 kg with a standard deviation of 0.025 kg .The random variable weight of the pack follows a normal distribution.
### a)What is the probability that the pack weighs less than 0.280kg?
### b)What is the probability that the pack weighs more than 0.350 kg?
### c)what is the probability that the pack weighs between 0.260 kg to 0.340kg?

```python
# calculate the z value  .Standerdized score.
z=(0.280 - 0.295)/0.025  # x= 0.280 ,standered deviation =0.025 mean =0.295
z
```

```python
# use the calculated z value to claculate the probability that the pack weighs less than 0.280kg
stats.norm.cdf(z) # norm.cdf normal cumulative distribution function
```

```python
# another way of calculating probability that the pack weighs less than 0.280kg
stats.norm.cdf(0.280,loc=0.295,scale=0.025) #loc = mean scale = standard deviation
```

```python
#probability that the pack weighs more than 0.350 kg  
#The probability (area) to the right is calculated as (1-probability to the left)
1-stats.norm.cdf(0.350,loc=0.295,scale=0.025)
```

```python
#probability that the pack weighs between 0.260 kg to 0.340kg
stats.norm.cdf(0.340,loc=0.295,scale=0.025)- stats.norm.cdf(0.260,loc=0.295,scale=0.025)
```

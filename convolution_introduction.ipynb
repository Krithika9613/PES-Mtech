```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
patch1=np.ones((5,10))*10
patch2=np.ones((5,10))*180
```

```python
patch=np.concatenate((patch1,patch2))
```

```python
#plt.imshow(patch,cmap='gray')
```

```python
kernal=np.matrix('1 1 1;1 1 1; 1 1 1')
kernal
```

```python
patch
```

```python
plt.imshow(patch,cmap='gray')
```

```python
con_feat=cv2.filter2D(patch,-1,kernal)/9
```

```python
plt.imshow(con_feat,cmap='gray')
```
